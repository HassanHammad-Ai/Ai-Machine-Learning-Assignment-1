## Question 1
Ask the user for a string that contains both letters and numbers (e.g., "Py2th0on3"). Write a program that separates them into two lists: one for the letters and one for the numbers. Calculate the sum of the numbers and join the letters to form a word.

- Example Input: "Py2th0on3"
- Example Output: Word: "Python", Sum: 5

````python
text = input("Enter a string with letters and numbers: ")

letters = []
numbers = []

for ch in text:
    if ch.isalpha():
        letters.append(ch)
    elif ch.isdigit():
        numbers.append(int(ch))

word = "".join(letters)
total = sum(numbers)

print(f'Word: "{word}", Sum: {total}')
````

## Question 2
Ask a user to enter a password. Check if it is valid based on these rules: Length must be between 8 and 16 characters, and it must contain at least one uppercase letter and one special character (e.g., @, #, $).

- Example Input: "Guest@123"
- Example Output: "Password is Secure."

````python
import string

password = input("Enter a password: ")

valid_length = 8 <= len(password) <= 16
has_upper = any(ch.isupper() for ch in password)
has_special = any(ch in string.punctuation for ch in password)

if valid_length and has_upper and has_special:
    print("Password is Secure.")
else:
    print("Password is NOT secure:")
    if not valid_length:
        print("- Length must be between 8 and 16 characters.")
    if not has_upper:
        print("- It must contain at least one uppercase letter.")
    if not has_special:
        print("- It must contain at least one special character (@, #, $ ...).")
````

## Question 3
A store gives discounts based on the total price: Under $50 (No discount), $50 to $100 (10% discount), and Over $100 (20% discount). If the user is a "Member," they get an additional 5% off the final calculated price.

- Example Input: Price: 100, Member: Yes
- Example Output: Final Total: $85.5 (Note: 100 - 10% = 90; then 90 - 5% = 85.5)

````python
price = float(input("Enter the total price: "))
member = input("Are you a member? (Yes/No): ").strip().lower()

if price < 50:
    total = price
elif price <= 100:
    total = price - price * 0.10
else:
    total = price - price * 0.20

if member in ("yes", "y"):
    total = total - total * 0.05

print(f"Final Total: ${round(total, 2)}")
````

## Question 4
Ask the user to input two separate lists of names (comma-separated). Create a third list that contains only the names that appear in both lists.

- Example Input: List 1: "Ali, Ben, Sara", List 2: "Ben, Joy, Ali"
- Example Output: Common names: ['Ali', 'Ben']

````python
list1 = [name.strip() for name in input("Enter list 1 (comma-separated): ").split(",")]
list2 = [name.strip() for name in input("Enter list 2 (comma-separated): ").split(",")]

common = []
for name in list1:
    if name in list2 and name not in common:
        common.append(name)

print("Common names:", common)
````

## Question 5
Create a dictionary where keys are student names and values are lists of 3 test scores. Write a program that calculates the average for each student. If the average is > 90, print the student's name followed by "Honor Roll."

- Example Input: {'Sam': [95, 90, 92]}
- Example Output: Sam: Average 92.3 - Honor Roll

````python
students = {
    "Sam": [95, 90, 92],
    "Mona": [80, 85, 70],
    "Omar": [100, 98, 95],
}

for name, scores in students.items():
    average = sum(scores) / len(scores)
    if average > 90:
        print(f"{name}: Average {average:.1f} - Honor Roll")
    else:
        print(f"{name}: Average {average:.1f}")
````

## Question 6
Ask the user for two words of the same length. Create a new string by interleaving their characters (one from the first word, one from the second).

- Example Input: word1 = "abc", word2 = "xyz"
- Example Output: "axbycz"

````python
word1 = input("Enter the first word: ")
word2 = input("Enter the second word: ")

if len(word1) != len(word2):
    print("Error: the two words must have the same length.")
else:
    result = ""
    for i in range(len(word1)):
        result += word1[i] + word2[i]

    print(f'"{result}"')
````

## Question 7
Ask for a multi-digit number. Sum its digits. If the result is more than one digit, keep summing the digits until you get a single-digit result.

- Example Input: 98
- Example Output: Digital Root: 8 (9+8=17, 1+7=8)

````python
number = input("Enter a multi-digit number: ").strip()

if not number.isdigit():
    print("Error: please enter a positive whole number.")
else:
    steps = []
    while len(number) > 1:
        digits = [int(d) for d in number]
        total = sum(digits)
        steps.append("+".join(number) + "=" + str(total))
        number = str(total)

    print(f"Digital Root: {number}", end="")
    if steps:
        print(" (" + ", ".join(steps) + ")")
    else:
        print()
````

## Question 8
Given a list of numbers with many duplicates, return a list of tuples showing each unique element and its frequency (how many times it appeared).

- Example Input: [1, 1, 2, 3, 3, 3, 4]
- Example Output: [(1, 2), (2, 1), (3, 3), (4, 1)]

````python
numbers = [1, 1, 2, 3, 3, 3, 4]

counts = {}
for n in numbers:
    counts[n] = counts.get(n, 0) + 1

result = [(value, freq) for value, freq in counts.items()]

print(result)
````

## Question 9
Ask the user for a number $n$. Calculate the factorial (n!) using a loop, and then check if the resulting factorial is divisible by the sum of numbers from 1 to $n$.

- Example Input: 4
- Example Output: Factorial is 24, Sum is 10. 24 is not divisible by 10.

````python
n = int(input("Enter a number n: "))

factorial = 1
for i in range(1, n + 1):
    factorial *= i

total = sum(range(1, n + 1))

print(f"Factorial is {factorial}, Sum is {total}.", end=" ")
if total != 0 and factorial % total == 0:
    print(f"{factorial} is divisible by {total}.")
else:
    print(f"{factorial} is not divisible by {total}.")
````

## Question 10
Start with a dictionary of items and prices: {'apple': 0.5, 'bread': 2.0, 'milk': 1.5}. Allow the user to "Add" or "Remove" items from a cart. If they add an item not in the store, print an error. Finally, print the total bill.

- Example Input: Add 'apple', Add 'milk'
- Example Output: Total Bill: $2.0

Type `add apple`, `remove milk`, and so on. Type `done` to finish.

````python
store = {"apple": 0.5, "bread": 2.0, "milk": 1.5}
cart = []

print("Store items:", store)
print("Type 'add <item>' or 'remove <item>'. Type 'done' to finish.")

while True:
    command = input("> ").strip().lower()

    if command == "done":
        break

    parts = command.split(maxsplit=1)
    if len(parts) != 2 or parts[0] not in ("add", "remove"):
        print("Invalid command. Use: add <item> / remove <item> / done")
        continue

    action, item = parts[0], parts[1].strip("'\" ")

    if action == "add":
        if item in store:
            cart.append(item)
            print(f"Added {item}.")
        else:
            print(f"Error: '{item}' is not in the store.")
    else:  # remove
        if item in cart:
            cart.remove(item)
            print(f"Removed {item}.")
        else:
            print(f"Error: '{item}' is not in your cart.")

bill = sum(store[item] for item in cart)
print(f"Total Bill: ${bill}")
````

- Example Input: "Race car"

- Example Input: n=3
- Example Output:

```
1 1 1
2 2 2
```

## Question 13
In a list of numbers, find all "peaks." A peak is a number that is greater than the number to its left and the number to its right.

- Example Input: [1, 5, 2, 6, 3]
- Example Output: Peaks: [5, 6]

````python
numbers = [1, 5, 2, 6, 3]

peaks = []
for i in range(1, len(numbers) - 1):
    if numbers[i] > numbers[i - 1] and numbers[i] > numbers[i + 1]:
        peaks.append(numbers[i])

print("Peaks:", peaks)
````

## Question 14
Take a dictionary like {'A': 1, 'B': 2, 'C': 1} and flip it. If two keys have the same value, the new value should be a list of those keys.

- Example Input: {'A': 1, 'B': 2, 'C': 1}
- Example Output: {1: ['A', 'C'], 2: ['B']}

````python
original = {"A": 1, "B": 2, "C": 1}

flipped = {}
for key, value in original.items():
    if value not in flipped:
        flipped[value] = []
    flipped[value].append(key)

print(flipped)
````

## Question 15
Create a dictionary where the keys are the vowels found in a user's string and the values are lists of the indices where those vowels appear.

- Example Input: "banana"
- Example Output: {'a': [1, 3, 5]}

````python
text = input("Enter a string: ")

vowels = "aeiou"
result = {}

for index, ch in enumerate(text):
    if ch.lower() in vowels:
        result.setdefault(ch.lower(), []).append(index)

print(result)
````
